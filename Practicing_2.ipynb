**Word frequency counter**

```python
text = "Hello Hello how are you ?"
words = text.split()
word_freq = {}
for word in words:
  if word in word_freq:
    word_freq[word] += 1
  else:
    word_freq[word] = 1
print(word_freq)
```

**Find key with maximum value in dictionary**

```python
numbers = {"a": 1, "b": 2, "c": 3, "d": 4}
max_key = list(numbers.keys())[0]
max_value = numbers[max_key]
for key,value in numbers.items():   # We use items here to get keys and values both
  if value >= max_value:
     max_key = key
     max_value = value
print(f"The key with the maximum value is {max_key} with the values {max_value}")
```

**Convert list of tuples into dictionary**

```python
# Method 1
tuples = [("a",1),("b",2),("c",3),("d",4),("e",5)]
dictionary = {}
for key,values in tuples :
  dictionary[key] = values
print(dictionary)
```

**Function to check prime numbers**

```python
num = int(input("Enter a number : "))
if num < 2 :
  print(f"{num} is not a prime number")
else:
   for i in range(2,num):
      if num % i == 0:
       print(f"{num} The number is not prime")
       break
      else:
       print(f"{num} The number is prime")
```

**Find factorial of a number**

```python
num = int(input("Enter a number : "))
factorial = 1
for i in range(1,num+1):
  factorial *= i
print(f"The factorial of {num} is {factorial}")
```

**Fibonacci sequence (first n numbers)**

```python
num = int(input("Enter a number : "))
Fibonacci = []
a,b = 0,1
for i in range(0,num+1) :
  Fibonacci.append(a)
  c = a + b
  a = b
  b=c
print(Fibonacci)
```

**Contact Book **
* By use dictionary: names → phone numbers, with add/search/update/remove features

```python
contact = {}

def show_menu():
  print("\n-----Main Menu-----")
  print("1) Add New Contact")
  print("2) Search a Contact")
  print("3) Update a Contact")
  print("4) Remove a Contact")
  print("5) Show all Contact")
  print("6) Exit")


def add_new_contact():
  name = input("Enter a name : ")
  num  = input("Enter a phone number : ")
  contact[name.lower()] = num
  print("Contact added successfully")


def search_contact():
  name = input("Enter a name : ")
  if name.lower() in contact:
    print(f"The phone number of {name} is {contact[name.lower()]}")
  else:
    print("Contact not found")

def update_contact():
  name = input("Enter a name : ")
  if name.lower() in contact:
    num = input("Enter a new phone number : ")
    contact[name.lower()] = num
    print("Contact updated successfully")
  else:
    print("Contact not found")


def remove_contact():
  name = input("Enter a name : ")
  if name.lower() in contact:
    del contact[name.lower()]
    print("Contact removed successfully")
  else:
    print("Contact not found")

def show_all_contact():
  if len(contact) == 0:
    print("Contact book is empty")
  else:
    print("-----Contact Book-----")
    for name,num in contact.items():
      print(f"{name} : {num}")

def exit_program():
  print("Exiting the program")



while True :
     show_menu()
     Choice = input("Enter your choice : ")
     if Choice == '1':
       add_new_contact()
     elif Choice == '2':
       search_contact()
     elif Choice == '3':
       update_contact()
     elif Choice == '4':
       remove_contact()
     elif Choice == '5':
       show_all_contact()
     elif Choice == '6':
       exit_program()
       break
     else:
       print("Invalid choice. Please try again! ")
```
