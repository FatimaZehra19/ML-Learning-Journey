Below we are using **Upper** and **Lower** to change the string in **lower** or **upper** and then we use **isupper** or **islower** to find weather is it True or False

```python
phrase = "Learning Python"
print(phrase.upper().isupper())
```

To find how many characters are there in string we use **LEN** function

```python
phrase = "Learning Python"
print(len(phrase))
```

If i want to grab any function of the string then we use square brackets and then write the index of the character.The index will starts from Zero.

```python
phrase = "Learning Python"
print(phrase[0])
```

Index Function will give the information about the character position in the string.

```python
phrase = "Learning Python"
print(phrase.index("P"))
```

Replace will change the word.

```python
phrase = "Learning Python"
print(phrase.replace("Python","Skill"))
```
