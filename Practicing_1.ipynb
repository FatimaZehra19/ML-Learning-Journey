**Reverse a string**
We have 3 methods from which we can reverse a string.
* While loop
* for loop
* Python Slice

```python
# While loop Method
word = "Fatima"
reversed_word = ""
i = len(word) - 1    # i is the ndex of last letter
while i >=0 :
  reversed_word += word[i]
  i -= 1
print(reversed_word)
```

```python
# For Loop Method
word = "Fatima"
reversed_word = ""
for i in range (len(word)   -1 , -1 , -1) :
                 # (word) start:stop:step
  reversed_word += word[i]

print(reversed_word)
```

```python
word = "Fatima"
print(word[::-1])
```

Count vowels in a string

```python
text = "I am tyring to count vowels in my string through python"
vowels ="a,e,i,o,u,A,E,I,O,U"
count = 0
for letter in text :
   if letter in vowels :
    count+= 1

print("The vowels in my string are " + str(count))
```

**Find largest number in a list**

```python
num_list = [125,285,852,863,624,631,844]
largest_num = num_list[0]
for num in num_list :
  if num > largest_num :
    largest_num = num

print ("The largest number in the list is " + str(largest_num))
```

**Sum all numbers in a list**

```python
num_list = [125,285,854,363,644,631,844]
total_sum = 0
for num in num_list :
  total_sum += num

print("The sum of all numbers in the list is " + str(total_sum))
```

**Print multiplication table using a loop**

```python
num_list = [8,5,6,4,2,1,7,5,2,5,6,4]
num = 5
for i in num_list:
  print(str(num) + " x " + str(i) + " = " + str(num*i))
```

**Todays Mini Project**
 To-Do List App

```python
tasks = []

def show_menu():
  print("\n----TO DO APP----")
  print("1) Add a New Task")
  print("2) View all Tasks")
  print("3) Remove a Task")
  print("4) Exit")

def add_task(tasks) :
  task = input("Enter a new task: ")
  if task :
        tasks.append(task)
        print(" ✅ Task added!")
  else :
        print("❌ Empty task not added!")


def view_task(tasks) :
  if not tasks :
    print("No Tasks Yet.")
    return
  print("Enter Your Tasks")
  for i, t in enumerate(tasks, start= 1):  # enumerate is used to mentions the index number of list items. And start = 1 is used to stat list from 1 not from 0.
       print(f"{i}.{t}")


def remove_task(tasks) :
  if not tasks :
    print("Nothing to remove")
    return
  view_task(tasks)
  choice = input("Enter the task number you want to remove: ")
  try :
    idx = int(choice)
    if 1<= idx <= len(tasks) :
      removed = tasks.pop(idx - 1) # we use -1 here becsuse we starts our string from 1 but byDefault it is 0.
      print( f"🗑️ Removed = {removed}") # Corrected f-string
    else:
      print("⚠️ Invalid number. Please choose a listed task number.")

  except ValueError:
    print("⚠️ Invalid input. Please enter a number.")



# ===== Main loop =====
def main() :
  while True :
    show_menu() # Added parentheses to call the function
    choice = input("Enter your choice from(1-4): ")
    if choice == "1":
      add_task(tasks)
    elif choice == "2":
      view_task(tasks)
    elif choice == "3":
      remove_task(tasks)
    elif choice == "4":
      print("👋 Goodbye!")
      break
    else :
      print(" ⚠️ Invalid Choice. Please Enter From (1-4)")

main()
```
